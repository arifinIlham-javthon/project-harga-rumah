```python
# Lihat contoh nilai kolom target
print("=== Contoh price_idr ===")
print(df['price_idr'].head(10))
print("\nStatistik price_idr:")
print(df['price_idr'].describe())
```

```python
# Cek berapa baris yang harganya ekstrem
for treshold in [1e9, 1e10, 1e11, 1e12, 1e13]:
    count = (df['price_idr'] > treshold).sum()
    print(f"Harga > Rp {treshold:,.0f}: {count} baris")
```

```python
# Lihat persentil harga
percentiles = [0.5, 0.9, 0.95, 0.99, 0.995, 0.999, 1.0]
print(df['price_idr'].quantile(percentiles))
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Distribusi asli
axes[0].hist(df['price_idr'], bins=50, edgecolor='black')
axes[0].set_title('Distribusi Harga (Asli)')
axes[0].set_xlabel('Harga (IDR)')
axes[0].set_ylabel('Frekuensi')

# Plot 2: Distribusi setelah log-transform (biar kelihatan pola)
import numpy as np
df_nonzero = df[df['price_idr'] > 0]
axes[1].hist(np.log(df_nonzero['price_idr']), bins=50, edgecolor='black', color='orange')
axes[1].set_title('Distribusi Harga (log10)')
axes[1].set_xlabel('log10(Harga)')
axes[1].set_ylabel('Frekuensi')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
df_nonzero.boxplot(column='price_idr', ax=ax)
ax.set_title('Boxplot Harga Rumsh')
plt.show()
```
