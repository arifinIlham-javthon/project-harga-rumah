```python
# Notebook 02: Preprocessing Data
# Tujuan: Bersihkan dataset mentah sebelum masuk ke modeling

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Biar tampilan output lebih rapi
pd.set_option('display.max_columns', None)

print("Library siap!")
```

```python
# Baca dataset mentah
df = pd.read_csv("../data/indonesia_housing_dataset.csv")

# cek ukuran awal
print(f"Jumlah baris awal: {len(df)}")
print(f"Jumlah kolom awal: {df.shape[1]}")
print(f"\nKolom-kolomnya: {df.columns.tolist()}")
```

```python
# Buang kolom leakage (yang bikin model "curang")
kolom_leakage = ['price_per_m2_land', 'price_per_m2_building']
df = df.drop(columns=kolom_leakage)

print(f"Kolom setelah buang leakage: {df.shape[1]}")
print(f"Kolom tersisa: {df.columns.tolist()}")
```

```python
# Buang baris dengan price_idr = 0 (data error)
sebelum = len(df)
df = df[df['price_idr'] > 0]
sesudah = len(df)

print(f"Baris sebelum: {sebelum}")
print(f"Baris sesudah: {sesudah}")
print(f"Dibuang: {sebelum - sesudah} baris")
```

```python
# Hitung batas atas (perentil 99.5)
batas_atas = df['price_idr'].quantile(0.995)
print(f"Batas atas (persentil 99.5): Rp {batas_atas:,.0f}")

# Buang baris di atas batas
sebelum = len(df)
df = df[df['price_idr'] <= batas_atas]
sesudah = len(df)

print(f"\nBaris sebelum: {sebelum}")
print(f"Baris sesudah: {sesudah}")
print(f"Dibuang: {sebelum - sesudah} baris")
```

```python
# Verifikasi hasil pembersihan
print("=== Statistik price_idr setelah dibersihkan ===")
print(f"Jumlah baris: {len(df)}")
print(f"Median: Rp {df['price_idr'].median():,.0f}")
print(f"Mean:   Rp {df['price_idr'].mean():,.0f}")
print(f"Min:    Rp {df['price_idr'].min():,.0f}")
print(f"Max:    Rp {df['price_idr'].max():,.0f}")

print("\n=== Cek outlier ekstrem ===")
print(f"Harga > Rp 1 triliun: {(df['price_idr'] > 1e12).sum()} baris")
print(f"Harga > Rp 100 miliar: {(df['price_idr'] > 1e11).sum()} baris")
```

```python
# Cek missing values di data yang sudah bersih
print("=== Missing values ===")
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)

print(f"\nTotal kolom dengan missing: {len(missing)}")
```

```python
# Simpan dataset bersih
df.to_csv("../data/indonesia_housing_clean.csv", index=False)

print("Dataset bersih disimpan ke: data/indonesia_housing_clean.csv")
print(f"Total baris: {len(df)}")
print(f"Total kolom: {df.shape[1]}")
```
